```python
import math


class Vector:
    """
    A mathematical vector.
    """

    def __init__(self, values):
        """
        Create a vector.

        Example:
        Vector([1, 2, 3])
        """

        self._validate_values(values)
        self.values = values

    def add(self, other):
        """
        Add two vectors.
        """

        self._validate_vector(other)

        result = []

        for i in range(len(self.values)):
            result.append(self.values[i] + other.values[i])

        return Vector(result)

    def subtract(self, other):
        """
        Subtract two vectors.
        """

        self._validate_vector(other)

        result = []

        for i in range(len(self.values)):
            result.append(self.values[i] - other.values[i])

        return Vector(result)

    def scale(self, scalar):
        """
        Multiply vector by scalar.
        """

        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be a number.")

        result = []

        for value in self.values:
            result.append(value * scalar)

        return Vector(result)

    def magnitude(self):
        """
        Compute vector magnitude.
        """

        total = 0

        for value in self.values:
            total += value ** 2

        return math.sqrt(total)

    def dot(self, other):
        """
        Compute dot product.
        """

        self._validate_vector(other)

        total = 0

        for i in range(len(self.values)):
            total += self.values[i] * other.values[i]

        return total

    def normalize(self):
        """
        Normalize vector to unit length.
        """

        mag = self.magnitude()

        if mag == 0:
            raise ValueError("Cannot normalize zero vector.")

        result = []

        for value in self.values:
            result.append(value / mag)

        return Vector(result)

    def dimension(self):
        """
        Return vector dimension.
        """

        return len(self.values)

    def _validate_vector(self, other):
        """
        Validate vector compatibility.
        """

        if not isinstance(other, Vector):
            raise TypeError("Input must be a Vector.")

        if len(self.values) != len(other.values):
            raise ValueError("Vectors must have same dimensions.")

    def _validate_values(self, values):
        """
        Validate input values.
        """

        if not isinstance(values, list):
            raise TypeError("Vector must be initialized with a list.")

        if len(values) == 0:
            raise ValueError("Vector cannot be empty.")

        for value in values:
            if not isinstance(value, (int, float)):
                raise TypeError("Vector values must be numbers.")

    def __repr__(self):
        """
        Pretty print vector.
        """

        return f"Vector({self.values})"
```

```python
class Matrix:
    """
    A mathematical matrix.
    """

    def __init__(self, values):
        """
        Create a matrix.

        Example:
        Matrix([
            [1,2],
            [3,4]
        ])
        """

        self._validate_values(values)
        self.values = values

    def add(self, other):
        """
        Add two matrices.
        """

        self._validate_matrix(other)

        result = []

        for i in range(len(self.values)):
            row = []

            for j in range(len(self.values[0])):
                row.append(
                    self.values[i][j] + other.values[i][j]
                )

            result.append(row)

        return Matrix(result)

    def subtract(self, other):
        """
        Subtract two matrices.
        """

        self._validate_matrix(other)

        result = []

        for i in range(len(self.values)):
            row = []

            for j in range(len(self.values[0])):
                row.append(
                    self.values[i][j] - other.values[i][j]
                )

            result.append(row)

        return Matrix(result)

    def scale(self, scalar):
        """
        Multiply matrix by scalar.
        """

        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be a number.")

        result = []

        for row in self.values:
            new_row = []

            for value in row:
                new_row.append(value * scalar)

            result.append(new_row)

        return Matrix(result)

    def transpose(self):
        """
        Transpose matrix.
        """

        rows = len(self.values)
        cols = len(self.values[0])

        result = []

        for j in range(cols):
            row = []

            for i in range(rows):
                row.append(self.values[i][j])

            result.append(row)

        return Matrix(result)

    def multiply(self, other):
        """
        Matrix multiplication.
        """

        if not isinstance(other, Matrix):
            raise TypeError("Input must be a Matrix.")

        rows_a = len(self.values)
        cols_a = len(self.values[0])

        rows_b = len(other.values)
        cols_b = len(other.values[0])

        if cols_a != rows_b:
            raise ValueError(
                "Columns of first matrix must equal rows of second matrix."
            )

        result = []

        for i in range(rows_a):
            row = []

            for j in range(cols_b):

                total = 0

                for k in range(cols_a):
                    total += (
                        self.values[i][k]
                        * other.values[k][j]
                    )

                row.append(total)

            result.append(row)

        return Matrix(result)

    def multiply_vector(self, vector):
        """
        Matrix × Vector multiplication.
        """

        if not isinstance(vector, Vector):
            raise TypeError("Input must be a Vector.")

        rows = len(self.values)
        cols = len(self.values[0])

        if cols != vector.dimension():
            raise ValueError(
                "Matrix columns must match vector dimension."
            )

        result = []

        for row in self.values:

            total = 0

            for i in range(cols):
                total += row[i] * vector.values[i]

            result.append(total)

        return Vector(result)

    def shape(self):
        """
        Return matrix shape.
        """

        return (
            len(self.values),
            len(self.values[0])
        )

    def _validate_matrix(self, other):
        """
        Validate matrix compatibility.
        """

        if not isinstance(other, Matrix):
            raise TypeError("Input must be a Matrix.")

        if self.shape() != other.shape():
            raise ValueError(
                "Matrices must have same shape."
            )

    def _validate_values(self, values):
        """
        Validate matrix input.
        """

        if not isinstance(values, list):
            raise TypeError(
                "Matrix must be initialized with a list."
            )

        if len(values) == 0:
            raise ValueError(
                "Matrix cannot be empty."
            )

        row_length = len(values[0])

        for row in values:

            if not isinstance(row, list):
                raise TypeError(
                    "Each row must be a list."
                )

            if len(row) != row_length:
                raise ValueError(
                    "All rows must have same length."
                )

            for value in row:
                if not isinstance(value, (int, float)):
                    raise TypeError(
                        "Matrix values must be numeric."
                    )

    def __repr__(self):
        """
        Pretty print matrix.
        """

        return f"Matrix({self.values})"
```

```python
m1 = Matrix([
    [1, 2],
    [3, 4]
])

m2 = Matrix([
    [5, 6],
    [7, 8]
])

v = Vector([1, 2])

print(m1.add(m2))
print(m1.subtract(m2))
print(m1.scale(2))
print(m1.transpose())
print(m1.multiply(m2))
print(m1.multiply_vector(v))
print(m1.shape())
```
